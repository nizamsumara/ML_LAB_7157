```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler

df = pd.read_csv("/content/auto-mpg (1).csv");
#Convet horsepower to numeric, replacing any non-numeric values(Like '?') with NaN
df['horsepower'] = pd.to_numeric(df['horsepower'], errors='coerce')

#Fill missing values in horsepower by apply median
df['horsepower'] = df['horsepower'].fillna(df['horsepower'].median())

#Select numeric columns
X = df[['mpg', 'cylixnders', 'displacement', 'weight', 'acceleration', 'model year', 'origin']]

#Rescale data
scaler = MinMaxScaler()
rescaled_data = scaler.fit_transform(X)
print(rescaled_data)
```

```python
from sklearn.preprocessing import LabelEncoder
#Create LabelEncoder object
encoder = LabelEncoder()

#Encode categorical column
df['car name'] = encoder.fit_transform(df['car name'])

#Display encoded data
print(df['car name'].head())
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

#select numeric fetaures
X = df[['mpg', 'cylinders', 'displacement', 'horsepower', 'weight', 'acceleration', 'model year']]

#handle missing values
X = X.fillna(X.mean())

#Standardize data
X_scaled = StandardScaler().fit_transform(X)

#apply PCA
pca = PCA(n_components=2)
principal_components = pca.fit_transform(X_scaled)


plt.scatter(principal_components[:,0], principal_components[:,1])
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('PCA Analysis of auto mgp dataset')
plt.show()
```
