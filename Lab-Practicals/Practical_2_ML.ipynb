```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler

#Convert horsepower to numeric, replacing any non-numeric values (like '?') with
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')

#Fill any missing value in horsepower with the median value
df['Age'] = df['Age'].fillna(df['Age'].median())

#Select numeric columns
X= df[['Age']]

#Rescale data
scaler = MinMaxScaler()
rescaled_data = scaler.fit_transform(X)


#Display result
print(rescaled_data)
```

## Additional Data Cleaning: Removing Null Data
The dataset currently has no missing values (as shown in the previous output). The following code removes rows containing null values if any are present and displays the updated shape.

```python
# Remove rows containing null values
df_cleaned = df.dropna().copy()

print("Missing Values After Removing Null Data:")
print(df_cleaned.isnull().sum())
print("\nShape Before Removing Null Data:", df.shape)
print("Shape After Removing Null Data:", df_cleaned.shape)
```

## Encoding Data
Encode selected categorical columns using one-hot encoding. The selected columns are Gender, Airport Continent, and Flight Status. Other text columns such as passenger names and IDs are identifiers or high-cardinality fields, so they are not encoded here.

```python
import pandas as pd

# Select useful categorical columns for encoding
categorical_features = ['Gender', 'Airport Continent', 'Flight Status']

# Apply one-hot encoding
encoded_data = pd.get_dummies(
    df_cleaned,
    columns=categorical_features,
    dtype=int
)

# Display encoded data
print(encoded_data.head())
print("\nShape after encoding:", encoded_data.shape)
```

## Feature Engineering and Feature Selection
Departure Date is converted into year, month, and day features. Then SelectKBest with the ANOVA F-test selects useful numerical features for predicting Flight Status.

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.feature_selection import SelectKBest, f_classif

# Create date-based numerical features
feature_data = df_cleaned.copy()
feature_data['Departure Date'] = pd.to_datetime(
    feature_data['Departure Date'], errors='coerce'
)
feature_data['Departure_Year'] = feature_data['Departure Date'].dt.year
feature_data['Departure_Month'] = feature_data['Departure Date'].dt.month
feature_data['Departure_Day'] = feature_data['Departure Date'].dt.day

# Select numerical input features and fill any missing values
selected_input_features = ['Age', 'Departure_Year', 'Departure_Month', 'Departure_Day']
X_features = feature_data[selected_input_features].copy()
X_features = X_features.fillna(X_features.median())

# Encode the target column (Flight Status)
target_encoder = LabelEncoder()
y_target = target_encoder.fit_transform(feature_data['Flight Status'].astype(str))

# Select the best numerical features
selector = SelectKBest(score_func=f_classif, k='all')
X_selected = selector.fit_transform(X_features, y_target)
selected_feature_names = X_features.columns[selector.get_support()].tolist()

print("Selected Features:")
print(selected_feature_names)
print("\nFeature Scores:")
print(pd.DataFrame({
    'Feature': X_features.columns,
    'Score': selector.scores_
}).sort_values(by='Score', ascending=False))
```

## Dimensionality Reduction using Principal Component Analysis (PCA)
PCA reduces the selected numerical features into two principal components. The data is standardized before applying PCA.

```python
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Standardize selected numerical features
scaler_pca = StandardScaler()
X_scaled_pca = scaler_pca.fit_transform(X_features[selected_feature_names])

# Apply PCA to reduce the features to two components
pca = PCA(n_components=2)
principal_components = pca.fit_transform(X_scaled_pca)

# Create a DataFrame for the principal components
pca_df = pd.DataFrame(
    principal_components,
    columns=['Principal Component 1', 'Principal Component 2']
)

print("PCA Result:")
print(pca_df.head())
print("\nExplained Variance Ratio:")
print(pca.explained_variance_ratio_)
print("Total Explained Variance:", pca.explained_variance_ratio_.sum())

# Visualize the two principal components
plt.figure(figsize=(8, 5))
plt.scatter(
    pca_df['Principal Component 1'],
    pca_df['Principal Component 2'],
    alpha=0.4
)
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('PCA - Airline Dataset')
plt.show()
```
